Creating a 6×8 Image

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
im = np.zeros((6,8), dtype=np.uint8)
im[2,3] = 255
fig, ax = plt.subplots(1, 1, figsize=(6, 8))
ax.imshow(im, cmap='gray', vmin=0, vmax=255)
ax.xaxis.set_ticks_position('top')
plt.show()

```

```python

im = np.zeros((6,8,3), dtype=np.uint8)
im[2,3] = (0, 255, 0)
fig, ax = plt.subplots(1, 1, figsize=(6, 8))
ax.imshow(im)
ax.xaxis.set_ticks_position('top')
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
im = cv.imread('C:\EN3160 - Image Processing and Machine Vision\L02\pic\poto.jpeg')
fig, ax = plt.subplots(1, 1, figsize=(6, 8))
ax.imshow(cv.cvtColor(im, cv.COLOR_BGR2GRAY), cmap='gray', vmin=0, vmax=255)
ax.xaxis.set_ticks_position('top')
plt.show()
```
